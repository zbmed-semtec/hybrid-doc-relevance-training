In case of reduction to compute the documents' WMD scores, we need to input pre-annotated documents'tokens to the code generate_wmd_similarity_with_reduction.py.

By doing this we restrict the documents' words to be either MeSH terms or non-MeSH terms, hence not both.

However, the prefix is used to annotate MeSH terms can be different from file to file.

In this repository, we employ the prefixes MeSHD or MeSHQ, as demonstrated in identifiers like MeSHD016424 or MeSHQ000628 for instance (see ../data/dic_MeShIDtoPMID_2022628.tsv). Therefore, the embeddings generated for MeSH terms will be stored using keys with the prefixes MeSHD or MeSHQ.

Hence, caution is advised when employing pre-annotated tokens as input for generating Word Mover's Distance (WMD). It is crucial to verify that the prefixes used for annotation match those utilized in this repository. Otherwise, the annotated words may not be recognized by the trained word2vec model, and the resulting embeddings for MeSH terms will not contribute to the computation of WMD scores.

To accomplish this, you can utilize this notebook for a quick verification.

```python
#To check the used prefixes
import numpy as np

pmids = []
docs = []

input_tokens = np.load('../data/RELISH/Tokenized_Input/RELISH_Annot_Tokens_Sample.npy', allow_pickle=True)

for line in range(len(input_tokens)):
        pmids.append(int(input_tokens[line][0]))
        docs.append(np.ndarray.tolist(input_tokens[line][1]))
        docs[line].extend(np.ndarray.tolist(input_tokens[line][2]))
        
print(docs[0])
```
